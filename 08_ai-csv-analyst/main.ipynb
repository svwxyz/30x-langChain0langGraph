```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
import pandas as pd
from dotenv import load_dotenv
load_dotenv()
import os 

api = os.environ['OPENAI_API_KEY']

if api:
    print('api set')
else:
    print('not set')
```

```python
llm = ChatOpenAI(model='gpt-5-mini')
```

```python
df = pd.read_csv('Data/data.csv')
```

```python
prompts = ChatPromptTemplate.from_messages([

        ('system','You are my Senior Data Analyzer who analyze csv and report to me according to my needs'),
        ('human',"""
                        This is my query 
                            {query}
                        the csv which you have to analyze is this 
                            {data}
                """)
])
```

```python
chain = prompts | llm 
```

```python
theinput = input("Enter what you want to ask about sales")
```

```python
response = chain.invoke({
        'query':theinput,
        'data':df.to_string(index=False)
}).content
```
